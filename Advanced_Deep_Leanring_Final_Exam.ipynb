## Q1: One-Hot Encoding

```python
from sklearn.preprocessing import OneHotEncoder

words = [["cat"], ["dog"], ["bird"], ["cat"]]

encoder = OneHotEncoder()

encoded = encoder.fit_transform(words)

print(encoded.toarray())
```

## Q2: TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "I love NLP",
    "NLP is amazing",
    "Machine learning is fun"
]

vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(documents)

print(tfidf_matrix.toarray())
```

## Q3: CountVectorizer

```python
from sklearn.feature_extraction.text import CountVectorizer

docs = ["apple orange apple","orange banana"]

vectorizer = CountVectorizer()

matrix = vectorizer.fit_transform(docs)

print(matrix.toarray())
```

## Q11: Stopword Removal

```python
from nltk.corpus import stopwords
nltk.download('stopwords')

words = ["I", "love", "to", "learn", "NLP"]

stop_words = set(stopwords.words("english"))

filtered = [word for word in words if word.lower() not in stop_words]

print(filtered)
```

## Q12: Stemming

```python
from nltk.stem import PorterStemmer
#nltk.download('porter')
stemmer = PorterStemmer()

words = ["playing", "played", "plays"]

stems = [stemmer.stem(word) for word in words] # used to stem function

print(stems)
```

## Q13: Lemmatization

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
lemmatizer = WordNetLemmatizer()

words = ["running", "better", "cars"]

lemmas = [lemmatizer.lemmatize(word) for word in words]

print(lemmas)
```

## Q15: Remove HTML

```python
import re

html = "<h1>Welcome</h1><p>This is NLP.</p>"

#write rest of the code
cleans = re.sub(r"<.*?>", " ", html)
clean = re.sub(r"\s+", " ", cleans)
print(clean.lower())
```

## Q16: Bag of Words
